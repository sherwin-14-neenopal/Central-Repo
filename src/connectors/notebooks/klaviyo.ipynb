```python
import os
import json
import pandas as pd


from typing import Optional
from typing import Dict, Any
from dotenv import load_dotenv
from klaviyo_api import KlaviyoAPI
```

```python
# Loading api token from config dir
load_dotenv(os.path.join("..", "..", "config", ".env"))
API_KEY_KLAVIYO = os.environ.get("API_KEY_KLAVIYO")

klaviyo = KlaviyoAPI(API_KEY_KLAVIYO, max_delay = 60, max_retries = 3, test_host = None)  
```

```python
## The function show_json would help prettifying JSON outputs

def show_json(variable: Dict[str, Any]) -> None:
    print(json.dumps(variable, indent = 2))
```

### Profiles

```python
try:
    profiles = klaviyo.Profiles.get_profiles(
                additional_fields_profile =  ['predictive_analytics'], 
                fields_profile =  ['first_name', 'location'],
                filter = None, 
                page_cursor = None,
                page_size = 20, 
                sort = 'created'
    )
    show_json(profiles)

except Exception as e:
    print(f"An error occurred: {e}")
```

```python
try:
        lists_for_profile = klaviyo.Profiles.get_lists_for_profile (
                    id = '01JNT5GR3ZVQEW3QVTXKCRN4XD', 
                    fields_list = ['name', 'opt_in_process']
        )

        show_json(lists_for_profile)

except Exception as e:
    print(f"An error occurred: {e}")
```

```python
try:
    segements_for_profile = klaviyo.Profiles.get_segments_for_profile (
                id ='01JNT5GR3ZVQEW3QVTXKCRN4XD', 
                fields_segment = ['name', 'is_active']
    )
    show_json(segements_for_profile)

except Exception as e:
    print(f"An error occurred: {e}")
```

```python
try:
    bulk_import_profiles = klaviyo.Profiles.get_bulk_import_profiles_jobs(
        fields_profile_bulk_import_job = ['status', 'total_count'],
        filter = None,
        page_cursor = None,
        page_size = 20,
        sort = 'created_at'
    )
    show_json(bulk_import_profiles)

except Exception as e:
    print(f"An error occurred: {e}")
```

### Reviews

```python
try:
    reviews = klaviyo.Reviews.get_reviews(
        fields_event = ['timestamp','event_properties'], 
        fields_review = ['email'], 
        filter = None,
        include = ['events'],
        page_cursor = None,
        page_size = 20,
        sort = 'rating'
    )
    show_json(reviews)

except Exception as e:
    print(f"An error occurred: {e}")
```

```python
try:
    categories = klaviyo.Catalogs.get_catalog_categories(
              fields_catalog_category = ['external_id','name'],
              filter = None ,
              page_cursor = None,
              sort = 'created',
    )
    show_json(categories)

except Exception as e:
    print(f"An error occurred: {e}")
```
