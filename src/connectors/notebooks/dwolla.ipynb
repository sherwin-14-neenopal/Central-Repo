```python
import os
import json
import dwollav2
import pandas as pd


from typing import Optional
from typing import Dict, Any
from dotenv import load_dotenv
```

```python
load_dotenv(os.path.join("..", "..", "config", ".env"))

client = dwollav2.Client(
  key = os.environ['DWOLLA_KEY'],
  secret = os.environ['DWOLLA_SECRET'],
  environment = 'sandbox'
)

token = client.Token(access_token = os.environ['ACCESS_TOKEN_DWOLLA'],expires_in = 3600)
```

```python
def show_json(variable: Dict[str, Any]) -> None:
    print(json.dumps(variable, indent = 2))
```

### Customers

```python
try:
    customers = token.get('https://api-sandbox.dwolla.com/customers/395aaa1c-d7d0-4b1c-a425-8497f03039a8')
    show_json(customers.body)

except Exception as e:
    print(f'An error occurred: {e}')
```
